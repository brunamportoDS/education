## Data Processing: Student Demographics

### Objective
Calculate the percentage of minority students for each school from raw enrollment data.

### Processing Steps
The CCD membership file contains over 12 million records with student counts broken down by:
- School (NCESSCH)
- Grade level
- Race/ethnicity (7 categories)
- Sex (Male/Female)

To efficiently process this large file, we:
1. **Read in chunks** (100,000 rows at a time) to manage memory
2. **Filter** to only schools in our analysis (from EdGap dataset)
3. **Aggregate** counts across all grades and sexes for each school
4. **Calculate** percentage minority as: (1 - White students / Total students) × 100

This approach reduces 12+ million rows to ~97,000 school-level records.

```python
# === Load school information with only needed columns ===
print("Loading school directory data...")

# Select only the columns you need
needed_school_cols = [
    "NCESSCH",  # join key
    "SCH_NAME",  # school name
    "LEA_NAME",  # district name
    "STATENAME",  # state name
    "LSTATE",  # state abbreviation
    "LZIP",  # zip code
    "LATCOD",  # latitude
    "LONCOD",  # longitude
    "SCHOOL_TYPE_TEXT",  # school type
    "LEVEL",  # school level (elementary, middle, high)
]

school_information = pd.read_csv(
    DIR_PATH,
    dtype=str,
    encoding="unicode_escape",
    usecols=lambda col: col in needed_school_cols,
    low_memory=False,
)

print(f"✓ Loaded school information: {school_information.shape}")
print(f"  Columns: {school_information.columns.tolist()}")
print("\nFirst few rows:")
print(school_information.head(3))
```

```python
# === Filter to regular schools to reduce dataset size ===
print(f"Original school count: {len(school_information)}")

if "SCHOOL_TYPE_TEXT" in school_information.columns:
    school_information = school_information[
        school_information["SCHOOL_TYPE_TEXT"].str.contains("Regular", case=False, na=False)
    ]
    print(f"✓ After filtering to regular schools: {len(school_information)}")
else:
    print("⚠️ SCHOOL_TYPE_TEXT column not found, skipping filter")

# Clean NCESSCH
school_information["NCESSCH"] = school_information["NCESSCH"].astype(str).str.strip()
```

```python
# === Load membership file - MEMORY OPTIMIZED ===
print("\nLoading membership data (optimized for memory)...")

# Get list of schools we need
need_ids = set(school_information["NCESSCH"].dropna())
print(f"Number of schools to match: {len(need_ids)}")

needed_cols = [
    "NCESSCH",
    "RACE_ETHNICITY",
    "STUDENT_COUNT",
]

print(f"Reading columns: {needed_cols}")

# Process in chunks with filtering
filtered_chunks = []
chunk_count = 0
total_rows_processed = 0

for chunk in pd.read_csv(
    MEM_PATH,
    usecols=needed_cols,
    dtype={"NCESSCH": str, "RACE_ETHNICITY": str, "STUDENT_COUNT": str},
    encoding="unicode_escape",
    engine="python",
    on_bad_lines="skip",
    chunksize=100_000,
):
    total_rows_processed += len(chunk)

    # Filter to only schools we care about
    chunk = chunk[chunk["NCESSCH"].isin(need_ids)]

    # Filter out "Not Specified" and "No Category Codes"
    chunk = chunk[~chunk["RACE_ETHNICITY"].isin(["Not Specified", "No Category Codes"])]

    # Convert STUDENT_COUNT to numeric immediately
    chunk["STUDENT_COUNT"] = pd.to_numeric(chunk["STUDENT_COUNT"], errors="coerce")

    # Drop rows with no student count or zero
    chunk = chunk[chunk["STUDENT_COUNT"].notna() & (chunk["STUDENT_COUNT"] > 0)]

    if not chunk.empty:
        filtered_chunks.append(chunk)
        chunk_count += 1
        if chunk_count % 10 == 0:
            print(f"  Processed {chunk_count} chunks ({total_rows_processed:,} rows scanned)...")

if filtered_chunks:
    membership_raw = pd.concat(filtered_chunks, ignore_index=True)
else:
    membership_raw = pd.DataFrame(columns=needed_cols)

print(f"\n✓ Filtered membership data loaded: {membership_raw.shape}")
print(f"  Scanned: {total_rows_processed:,} rows")
print(f"  Kept: {len(membership_raw):,} rows ({len(membership_raw)/total_rows_processed*100:.1f}%)")
print(f"\nMemory usage: {membership_raw.memory_usage(deep=True).sum() / 1024**2:.1f} MB")

print("\nSample of filtered data:")
print(membership_raw.head(10))

print("\nUnique RACE_ETHNICITY values:")
print(membership_raw["RACE_ETHNICITY"].value_counts())

print("\nStudent count statistics:")
print(membership_raw["STUDENT_COUNT"].describe())
```

```python
# === Clean NCESSCH ===
membership_raw["NCESSCH"] = membership_raw["NCESSCH"].astype(str).str.strip()

print(f"Processing {len(membership_raw):,} rows...")

# === Calculate total enrollment per school ===
# Sum across all races (and implicitly across grades/sex since we have multiple rows per school)
print("\nCalculating total enrollment...")

total_enrollment = membership_raw.groupby("NCESSCH", as_index=False)["STUDENT_COUNT"].sum()
total_enrollment.columns = ["NCESSCH", "total_enrollment"]

print(f"✓ Total enrollment calculated for {len(total_enrollment):,} schools")
print(f"  Total students across all schools: {total_enrollment['total_enrollment'].sum():,.0f}")
print("\nSample enrollments:")
print(total_enrollment.head(10))

# === Calculate counts by race/ethnicity ===
# This will sum across all grades and sexes for each race category per school
print("\nCalculating race/ethnicity counts...")

race_counts = membership_raw.groupby(["NCESSCH", "RACE_ETHNICITY"], as_index=False)["STUDENT_COUNT"].sum()

print(f"✓ Race counts aggregated: {race_counts.shape}")
print("\nSample race counts:")
print(race_counts.head(15))

# === Pivot to wide format ===
print("\nPivoting to wide format (one column per race)...")

race_wide = race_counts.pivot(
    index="NCESSCH", columns="RACE_ETHNICITY", values="STUDENT_COUNT"
).reset_index()

# Fill NaN with 0 (means that race category had no students at that school)
race_wide = race_wide.fillna(0)

print(f"\n✓ Pivoted race data: {race_wide.shape}")
print(f"  Race/ethnicity columns: {[c for c in race_wide.columns if c != 'NCESSCH']}")
print("\nSample pivoted data:")
print(race_wide.head())
print("\nColumn statistics:")
print(race_wide.describe())

# === Calculate percentage minority ===
print("\nCalculating percentage minority...")
# Minority students are all students who are not White
race_wide['minority_count'] = race_wide.drop(columns=['NCESSCH', 'White']).sum(axis=1)

# Merge with total enrollment to calculate percentage
membership_aggregated = total_enrollment.merge(race_wide[['NCESSCH', 'minority_count']], on='NCESSCH', how='left')
membership_aggregated['pct_minority'] = (membership_aggregated['minority_count'] / membership_aggregated['total_enrollment']) * 100

print("✓ Percentage minority calculated")
print("\nSample membership_aggregated data:")
print(membership_aggregated.head())
```

For each key predictor, we:

3. **Evaluate performance** using R², RMSE, and MAE
4. **Check residuals** to assess model fit

### Predictors to Examine

- **Percentage minority students** (additional predictor)

This step-by-step approach helps us understand each variable's contribution before combining them in a multiple regression model.

```python
# === Single Input Model: Median Income ===

# Plot regression line and scatter plot
plt.figure(figsize=(6, 6))

sns.regplot(
    data=final_data,
    x='ct_median_household_income',
    y='school_act_avg',
    color='blue',
    ci=False,
    scatter_kws={'color': 'black', 'edgecolors': 'white', 'linewidths': 1},
)

# Add axis labels
plt.xlabel('Median Household Income ($)', fontsize=16)
plt.ylabel('Average ACT Score', fontsize=16)

# Increase the fontsize of the tick labels
plt.tick_params(labelsize=14)

plt.show()
```

```python
# === Fit Linear Regression Model ===
import statsmodels.formula.api as smf

# Fit the simple linear regression model
model_median_income = smf.ols(
    formula='school_act_avg ~ ct_median_household_income', data=final_data
).fit()

# Display model summary
print(model_median_income.summary())

# Compute R-squared
print("\n" + "=" * 60)
print("MODEL METRICS")
print("=" * 60)
print(f"R-squared: {model_median_income.rsquared:.4f}")
print(
    f"This means that {model_median_income.rsquared*100:.2f}% of the variance in ACT scores"
)
print(f"is explained by median household income alone.")
```

```python
from sklearn.model_selection import KFold, cross_validate
from sklearn.linear_model import LinearRegression
import numpy as np

# 1. Define predictor and target variables
# Using the reduced model predictors and the cleaned data
predictor_variables = [
    'ct_pct_adults_college_degree',
    'school_pct_free_reduced_lunch',
    'pct_minority',
    # Add feature engineered columns if desired for cross-validation
    # 'college_lunch_interaction',
    # 'lunch_minority_interaction',
    # 'school_pct_free_reduced_lunch_sq'
]

target_variable = 'school_act_avg'

# Select the relevant columns
data_for_cv = final_data_cleaned[predictor_variables + [target_variable]].copy()

# 2. Handle any remaining missing values
print(f"Shape before dropping NaNs: {data_for_cv.shape}")
data_for_cv.dropna(inplace=True)
print(f"Shape after dropping NaNs: {data_for_cv.shape}")

X = data_for_cv[predictor_variables]
y = data_for_cv[target_variable]

# 3. Initialize KFold
n_splits = 5 # You can adjust the number of splits
kf = KFold(n_splits=n_splits, shuffle=True, random_state=42)

# 4. Initialize Linear Regression model
model_cv = LinearRegression()

# 5. Use cross_validate to perform k-fold cross-validation
scoring = {
    'r2': 'r2',
    'neg_mean_squared_error': 'neg_mean_squared_error',
    'neg_mean_absolute_error': 'neg_mean_absolute_error'
}

cv_results = cross_validate(
    model_cv,
    X,
    y,
    cv=kf,
    scoring=scoring,
    return_train_score=True # Optional: return train scores as well
)

# 6. Store the results of the cross-validation
print("\nCross-validation results:")
# Convert negative scores to positive for easier interpretation
cv_results_positive = {
    'test_r2': cv_results['test_r2'],
    'test_rmse': np.sqrt(-cv_results['test_neg_mean_squared_error']),
    'test_mae': -cv_results['test_neg_mean_absolute_error'],
    'train_r2': cv_results['train_r2'],
    'train_rmse': np.sqrt(-cv_results['train_neg_mean_squared_error']),
    'train_mae': -cv_results['train_neg_mean_absolute_error'],
    'fit_time': cv_results['fit_time'],
    'score_time': cv_results['score_time']
}

# Print mean and standard deviation of the results
print(f"Test R²: Mean = {np.mean(cv_results_positive['test_r2']):.4f}, Std = {np.std(cv_results_positive['test_r2']):.4f}")
print(f"Test RMSE: Mean = {np.mean(cv_results_positive['test_rmse']):.4f}, Std = {np.std(cv_results_positive['test_rmse']):.4f}")
print(f"Test MAE: Mean = {np.mean(cv_results_positive['test_mae']):.4f}, Std = {np.std(cv_results_positive['test_mae']):.4f}")
print(f"Train R²: Mean = {np.mean(cv_results_positive['train_r2']):.4f}, Std = {np.std(cv_results_positive['train_r2']):.4f}")
print(f"Train RMSE: Mean = {np.mean(cv_results_positive['train_rmse']):.4f}, Std = {np.std(cv_results_positive['train_rmse']):.4f}")
print(f"Train MAE: Mean = {np.mean(cv_results_positive['train_mae']):.4f}, Std = {np.std(cv_results_positive['train_mae']):.4f}")
```

```python
# === Load membership file - MEMORY OPTIMIZED ===
print("\nLoading membership data (optimized for memory)...")

# Get a set of NCESSCH IDs from the school information DataFrame to filter the large membership file
need_ids = set(school_information["NCESSCH"].dropna())
print(f"Number of schools to match: {len(need_ids)}")

# Define the necessary columns from the membership file
needed_cols = [
    "NCESSCH", # School ID for joining
    "RACE_ETHNICITY", # Student race/ethnicity category
    "STUDENT_COUNT", # Number of students in this category
]

print(f"Reading columns: {needed_cols}")

# Initialize a list to store filtered data chunks
filtered_chunks = []
chunk_count = 0
total_rows_processed = 0

# Read the large membership CSV file in chunks to manage memory
# usecols loads only specified columns, dtype ensures NCESSCH and RACE_ETHNICITY are strings
# encoding="unicode_escape" handles encoding issues, engine="python" is used for on_bad_lines
# on_bad_lines='skip' skips rows that cause parsing errors
# chunksize determines the number of rows read at a time
for chunk in pd.read_csv(
    MEM_PATH,
    usecols=needed_cols,
    dtype={"NCESSCH": str, "RACE_ETHNICITY": str, "STUDENT_COUNT": str},
    encoding="unicode_escape",
    engine="python",
    on_bad_lines="skip",
    chunksize=100_000,
):
    # Keep track of the total number of rows processed
    total_rows_processed += len(chunk)

    # Filter the current chunk to keep only rows for the schools in our need_ids set
    chunk = chunk[chunk["NCESSCH"].isin(need_ids)]

    # Filter out rows where RACE_ETHNICITY is 'Not Specified' or 'No Category Codes'
    chunk = chunk[~chunk["RACE_ETHNICITY"].isin(["Not Specified", "No Category Codes"])]

    # Convert 'STUDENT_COUNT' to numeric, coercing errors to NaN
    chunk["STUDENT_COUNT"] = pd.to_numeric(chunk["STUDENT_COUNT"], errors="coerce")

    # Drop rows with no student count or zero
    chunk = chunk[chunk["STUDENT_COUNT"].notna() & (chunk["STUDENT_COUNT"] > 0)]

    # If the filtered chunk is not empty, append it to the list
    if not chunk.empty:
        filtered_chunks.append(chunk)
        chunk_count += 1
        # Print progress every 10 chunks
        if chunk_count % 10 == 0:
            print(f"  Processed {chunk_count} chunks ({total_rows_processed:,} rows scanned)...")

# Concatenate all filtered chunks into a single DataFrame if any chunks were processed
if filtered_chunks:
    membership_raw = pd.concat(filtered_chunks, ignore_index=True)
else:
    # Create an empty DataFrame with the needed columns if no chunks were processed
    membership_raw = pd.DataFrame(columns=needed_cols)

print(f"\n✓ Filtered membership data loaded: {membership_raw.shape}")
print(f"  Scanned: {total_rows_processed:,} rows")
# Calculate and print the percentage of rows kept
print(f"  Kept: {len(membership_raw):,} rows ({len(membership_raw)/total_rows_processed*100:.1f}%)")
# Print the memory usage of the resulting DataFrame
print(f"\nMemory usage: {membership_raw.memory_usage(deep=True).sum() / 1024**2:.1f} MB")

print("\nSample of filtered data:")
# Display the first 10 rows of the filtered membership data
print(membership_raw.head(10))

print("\nUnique RACE_ETHNICITY values:")
# Display the counts of unique values in the 'RACE_ETHNICITY' column
print(membership_raw["RACE_ETHNICITY"].value_counts())

print("\nStudent count statistics:")
# Display descriptive statistics for the 'STUDENT_COUNT' column
print(membership_raw["STUDENT_COUNT"].describe())
```

```python
# === Clean NCESSCH ===
# Ensure the NCESSCH column is string type and remove leading/trailing whitespace
membership_raw["NCESSCH"] = membership_raw["NCESSCH"].astype(str).str.strip()

print(f"Processing {len(membership_raw):,} rows...")

# === Calculate total enrollment per school ===
# Group the raw membership data by school (NCESSCH) and sum the student counts
# This gives the total enrollment for each school across all grades and races
print("\nCalculating total enrollment...")

total_enrollment = membership_raw.groupby("NCESSCH", as_index=False)["STUDENT_COUNT"].sum()
# Rename the aggregated column to 'total_enrollment'
total_enrollment.columns = ["NCESSCH", "total_enrollment"]

print(f"✓ Total enrollment calculated for {len(total_enrollment):,} schools")
print(f"  Total students across all schools: {total_enrollment['total_enrollment'].sum():,.0f}")
print("\nSample enrollments:")
# Display the first 10 rows of the total enrollment DataFrame
print(total_enrollment.head(10))

# === Calculate counts by race/ethnicity ===
# Group by school and race/ethnicity to get the student count for each race within each school
print("\nCalculating race/ethnicity counts...")

race_counts = membership_raw.groupby(["NCESSCH", "RACE_ETHNICITY"], as_index=False)["STUDENT_COUNT"].sum()

print(f"✓ Race counts aggregated: {race_counts.shape}")
print("\nSample race counts:")
# Display the first 15 rows of the race counts DataFrame
print(race_counts.head(15))

# === Pivot to wide format ===
# Reshape the race_counts DataFrame so that each race/ethnicity becomes a column
# The index is NCESSCH, columns are RACE_ETHNICITY values, and values are STUDENT_COUNT
print("\nPivoting to wide format (one column per race)...")

race_wide = race_counts.pivot(
    index="NCESSCH", columns="RACE_ETHNICITY", values="STUDENT_COUNT"
).reset_index()

# Fill any missing values (NaN) with 0.0, as NaN here indicates zero students of that race at that school
race_wide = race_wide.fillna(0)

print(f"\n✓ Pivoted race data: {race_wide.shape}")
# Print the names of the race/ethnicity columns
print(f"  Race/ethnicity columns: {[c for c in race_wide.columns if c != 'NCESSCH']}")
print("\nSample pivoted data:")
# Display the first few rows of the pivoted DataFrame
print(race_wide.head())
print("\nColumn statistics:")
# Display descriptive statistics for the pivoted race count columns
print(race_wide.describe())

# === Calculate percentage minority ===
print("\nCalculating percentage minority...")
# Calculate the total count of minority students by summing all race counts except 'White'
race_wide['minority_count'] = race_wide.drop(columns=['NCESSCH', 'White']).sum(axis=1)

# Merge with total enrollment to calculate percentage
membership_aggregated = total_enrollment.merge(race_wide[['NCESSCH', 'minority_count']], on='NCESSCH', how='left')
# Calculate the percentage of minority students: (minority count / total enrollment) * 100
membership_aggregated['pct_minority'] = (membership_aggregated['minority_count'] / membership_aggregated['total_enrollment']) * 100

print("✓ Percentage minority calculated")
print("\nSample membership_aggregated data:")
# Display the first few rows of the final aggregated membership data
print(membership_aggregated.head())
```
